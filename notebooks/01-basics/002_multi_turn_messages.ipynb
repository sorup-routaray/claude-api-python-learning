```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content" : text}
    messages.append(user_message)
```

```python
def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content" : text}
    messages.append(assistant_message)
```

```python
def get_messages_text(message):
    return next(b.text for b in message.content if b.type == "text")
```

```python
#Load .env and create the Claude client (src/utils/claude_api_client.py)
from utils import ClaudeClient

claude = ClaudeClient(max_tokens=100)
claude.model
```

```python
def chat(messages):
    message = claude.create(messages)
    return get_messages_text(message)
```

```python
#Make a starting list of messages
messages = []
# Add in the initial user question of "Define quantum computing in one sentence"
add_user_message(messages, "Define quantum computing in one sentence")

#Pass the list of messages into 'chat' to get answer
answer = chat(messages)

# Take the answer and add it as an assistant message to the list of messages
add_assistant_message(messages, answer)
messages
#Add in a follow-up user question of 
add_user_message(messages, "Write another sentence.")

answer = chat(messages)
answer
```
