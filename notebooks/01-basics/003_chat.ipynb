```python
#Create the Claude client (loads .env)
from utils import ClaudeClient

claude = ClaudeClient(max_tokens=100)
claude.model
```

```python
#Make an initial list of messages
messages = []

#Keep asking the user for input until they type 'exit'
while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    print(f"You: {user_input}")
    claude.add_user_message(messages, user_input)

    answer = claude.chat(messages)

    claude.add_assistant_message(messages, answer)
    print(f"Assistant: {answer}")
```
