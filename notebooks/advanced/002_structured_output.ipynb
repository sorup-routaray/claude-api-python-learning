````python
#Create an Claude client
from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-5-5"
````

````python
def add_user_message(messages, text):
    user_message = {"role": "user", "content" : text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content" : text}
    messages.append(assistant_message)

def get_messages_text(message):
    return next(b.text for b in message.content if b.type == "text")
````

````python
def chat(messages, system=None, max_tokens=1024, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": max_tokens,
        "messages": messages,
    }
    if system is not None:
        params["system"] = system
    if stop_sequences is not None:
        params["stop_sequences"] = stop_sequences
    

    message = client.messages.create(**params)
    return "".join(b.text for b in message.content if b.type == "text")
````

````python
messages = []

prompt = "Generate three different sample AWS CLI commands. Each should be very short"

add_user_message(messages, prompt)
#add_assistant_message(messages, "```bash")
#response = chat(messages, stop_sequences=["```"])
response = chat(messages)
response.strip()
````
