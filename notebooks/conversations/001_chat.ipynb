```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content" : text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content" : text}
    messages.append(assistant_message)

def get_messages_text(message):
    return next(b.text for b in message.content if b.type == "text")
```

```python
#Create an Claude client
from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-5-5"
```

```python
def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=100,
        messages=messages
    )
    return get_messages_text(message)
```

```python
#Makje an initial list of messages
messages = []

#Use a 'While true' loop to keep asking the user for input until they type 'exit'
while True:
    #Ask the user for input
    user_input = input("You: ")
    
    #If the user types 'exit', break the loop
    if user_input.lower() == "exit":
        break
    
    print(f"You: {user_input}")
    #Add the user's input as a message to the list of messages
    add_user_message(messages, user_input)
    
    #Pass the list of messages into 'chat' to get answer
    answer = chat(messages)
    
    #Take the answer and add it as an assistant message to the list of messages
    add_assistant_message(messages, answer)
    
    #Print the assistant's answer
    print(f"Assistant: {answer}")
```
