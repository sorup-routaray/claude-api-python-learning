```python
#Load environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
#Create an Claude client
from anthropic import Anthropic
client = Anthropic()
model = "claude-opus-5-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content" : text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content" : text}
    messages.append(assistant_message)

def get_messages_text(message):
    return next(b.text for b in message.content if b.type == "text")

def chat(messages, system=None):
    
    default_parms = {
        "model": model,
        "max_tokens": 100,
        "messages": messages
    }
    if system is not None:
        default_parms["system"] = system

    message = client.messages.create(**default_parms)
        
    return get_messages_text(message)
```

```python
messages = []

system_prompt = """You are a patient math tutor.
    Do not directly give the answer to the user. 
    Instead, guide them through the problem step by step."""

add_user_message(messages, "How do i slove 5x+3=2 for x?")

answer = chat(messages, system=system_prompt)

answer
```
